# TeleDrive v2 — Telegram → Google Drive
Run the cells top-to-bottom. Never paste API keys or OAuth JSON into any AI chat.

```python
# 1) Install dependencies and unpack the project
import os, zipfile, subprocess, sys
from pathlib import Path
from google.colab import files

if not Path('/content/teledrive').exists():
    print('Upload the TeleDrive package zip (python-package.zip):')
    up = files.upload()
    zip_name = next(iter(up))
    with zipfile.ZipFile(zip_name) as z:
        z.extractall('/content')
    # move nested folder if needed
    for cand in ['/content/python-package', '/content/teledrive-package']:
        if Path(cand).exists() and not Path('/content/teledrive').exists():
            os.rename(cand, '/content/teledrive_pkg')
            break

pkg = '/content/teledrive_pkg' if Path('/content/teledrive_pkg').exists() else '/content/python-package'
sys.path.insert(0, pkg)
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', '-r', f'{pkg}/requirements.txt'])
print('Ready.')
```

```python
# 2) Bootstrap: create folders, DB, logging, health check
from teledrive import bootstrap
print(bootstrap.run())
```

```python
# 3) Secure credential entry — nothing is logged
import getpass, os
os.environ['TELEGRAM_API_ID'] = getpass.getpass('Telegram API ID: ')
os.environ['TELEGRAM_API_HASH'] = getpass.getpass('Telegram API Hash: ')
print('Now upload your Google OAuth Desktop client JSON in the Drive tab of the UI.')
```

```python
# 4) Launch the Gradio UI
from teledrive import app
app.launch(share=False, inline=True)
```
